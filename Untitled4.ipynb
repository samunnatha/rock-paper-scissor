```python
import random
userchoice=int(input("enter user choice: 0 for rock ,1 for paper,2 for scissors."))
if userchoice>=3 or userchoice<0:
  print("invalid number")
else:
  computer_choice=random.randint(0,2)
  print("computer choose")
  print(computer_choice)
  if computer_choice==userchoice:
    print("it's draw")
  elif computer_choice==0 and userchoice==2:
    print("you win")
  elif computer_choice==2 and userchoice==0:
    print("you lose")
  elif computer_choice>userchoice:
    print("you lose")
  elif computer_choice<userchoice:
    print("you win")
```
